```python
import os
import pydicom
import cv2
import numpy as np

# Cấu hình đường dẫn
input_dir = r"D:\tranh\Documents\Study\DataMining\Ricord_Data\midrc_ricord_1a"
output_dir = r"D:\tranh\Documents\Study\DataMining\X-MIR-Reproduction\data\Covidx\create_ricord_dataset\ricord_images"
txt_file = r"D:\tranh\Documents\Study\DataMining\X-MIR-Reproduction\data\Covidx\create_ricord_dataset\ricord_data_set.txt"

# Tạo thư mục đích
os.makedirs(output_dir, exist_ok=True)
txt_lines = []
print("Đang giải mã và xử lý ảnh DICOM, vui lòng chờ...")

# Quét và xử lý
for root, dirs, files in os.walk(input_dir):
    for file in files:
        if file.endswith('.dcm'):
            dcm_path = os.path.join(root, file)
            
            # Lấy ID bệnh nhân
            path_parts = dcm_path.replace("\\", "/").split("/")
            patient_folder = ""
            for part in path_parts:
                if part.startswith("MIDRC-RICORD-1A-"):
                    patient_folder = part
                    break
                    
            if not patient_folder:
                continue

            try:
                # Đọc DICOM (đã có pylibjpeg hỗ trợ giải mã JPEG Lossless)
                ds = pydicom.dcmread(dcm_path)
                img = ds.pixel_array
                
                # Chuẩn hóa ảnh
                img = img - np.min(img)
                img = (img / np.max(img) * 255).astype(np.uint8)
                
                # Lưu PNG
                out_filename = f"{patient_folder}-{file.replace('.dcm', '.png')}"
                out_path = os.path.join(output_dir, out_filename)
                
                cv2.imwrite(out_path, img)
                txt_lines.append(out_filename)
                
            except Exception as e:
                pass # Bỏ qua im lặng các file lỗi định dạng để luồng chạy liên tục

# Ghi file txt
with open(txt_file, 'w') as f:
    for line in txt_lines:
        f.write(line + "\n")

print(f"\n[THÀNH CÔNG] Đã sơ chế {len(txt_lines)} ảnh và lưu file danh sách!")
```
